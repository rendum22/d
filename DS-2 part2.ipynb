```python
sorted_test_score = np.sort(df['test_scores'])
sorted_test_score
```

```python
mean = np.mean(sorted_test_score)
std = np.std(sorted_test_score)
```

```python
outliers = []
for x in sorted_test_score:
    z = (x - mean)/std
    #z = np.abs(z)
    if z>1.5 or z<-1.5:
        outliers.append(x)
print(outliers)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Drop missing test_scores
df_clean = df[df['test_scores'].notnull()].copy()

# Plot original distribution
sns.histplot(df_clean['test_scores'], kde=True, bins=10)
plt.title("Original Distribution of Test Scores")
plt.xlabel("Test Scores")
plt.ylabel("Frequency")
plt.show()

# Print original skewness
from scipy.stats import skew
print("Skewness (Original):", skew(df_clean['test_scores']))
```

```python
# Step 1: Remove missing values
df_clean = df[df['test_scores'].notnull()].copy()

# Step 2: Apply log transformationimport numpy as np
from sklearn.preprocessing import StandardScaler

# Log1p to reduce skew (handles zeros safely)
df_clean['test_scores_log'] = np.log10(df_clean['test_scores'])

# Z-score normalization
scaler = StandardScaler()
df_clean['test_scores_transformed'] = scaler.fit_transform(df_clean[['test_scores_log']])

df_clean['test_scores_log'] = np.log10(df_clean['test_scores'])

# Step 3: Apply z-score normalization
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_clean['test_scores_transformed'] = scaler.fit_transform(df_clean[['test_scores_log']])
```

```python
# Plot transformed distribution
sns.histplot(df_clean['test_scores_transformed'], kde=True, bins=10)
plt.title("Transformed Distribution of Test Scores")
plt.xlabel("Transformed Test Scores")
plt.ylabel("Frequency")
plt.show()

# Print skewness after transformation
print("Skewness (Transformed):", skew(df_clean['test_scores_transformed']))
```
